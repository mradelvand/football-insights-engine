Mission 1 : Load the season and find our match (DataFrames, filters)

```python
from statsbombpy import sb
import pandas as pd

matches = sb.matches(competition_id=2, season_id=27)
```

Mission 2: load events

```python
match_id=game.iloc[0]['match_id']
```

```python
events=sb.events(match_id=match_id)
events.shape
```

```python
events['type'].value_counts()
```

```python
Everton_shots=events[(events['type'] == "Shot") & (events['team'] == "Everton")]
len(Everton_shots)
```

```python
Arsenal_shots=events[(events['type'] == "Shot") & (events['team'] == "Arsenal")]
len(Arsenal_shots)
```

```python
Everton_Pass=events[(events['type'] == "Pass") & (events['team'] == "Everton")]
len(Everton_Pass)
```
