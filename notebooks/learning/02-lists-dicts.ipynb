```python
from dataclasses import dataclass
from typing import Dict, List

## A class is a model (blueprint) for match events.

@dataclass
class MatchEvent:
    
    # Type hints describe the expected data type.

    id: str
    match_id: int
    index: int
    period: int
    timestamp: str
    minute: int
    second: int
    event_type: str
    team_id: int
    player_id: int
    location: List[float]
    details: Dict

# Create an object (instance) from the MatchEvent class.

pass_event = MatchEvent(
    id="evt-500",
    match_id=303299,
    index=1,
    period=1,
    timestamp="00:02:52.438",
    minute=2,
    second=52,
    event_type="Pass",
    team_id=103,
    player_id=7713,
    location=[34.2, 41.5],
    details={"length": 40.33}
)

# Access an object's attributes using dot notation.
print(pass_event.event_type)
print(pass_event.location)
```
